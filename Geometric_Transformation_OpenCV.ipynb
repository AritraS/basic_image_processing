#### Warp Affine

```python
tx=1000
ty=0
M = np.float32([[1,0,tx],[0,1,ty]])
M
```

```python
rows, cols,_ = image.shape
new_image = cv2.warpAffine(image,M,(cols,rows))
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
```

```python
rows, cols,_ = image.shape
new_image = cv2.warpAffine(image,M,(cols+tx,rows+ty))
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
```

#### Rotation

```python
theta=45
M =cv2.getRotationMatrix2D(center=(rows//2,cols//2), angle=theta, scale=1)
new_image = cv2.warpAffine(image, M, (cols, rows))
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
```

### Mathematical Operations

```python
new_image = image+20
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
```

```python
new_image = image*5
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
```

```python
Noise = np.random.normal(0,30,(rows, cols, 3)).astype(np.uint8)
new_image = image+Noise
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB))
```
