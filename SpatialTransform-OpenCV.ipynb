```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
```

```python
def plot_image(image_1, image_2, title_1="Original Image", title_2="New Image"):
    plt.figure(figsize=(10,20))
    plt.subplot(1,2,1)
    plt.imshow(cv2.cvtColor(image_1, cv2.COLOR_BGR2RGB))
    plt.title(title_1)
    plt.subplot(1,2,2)
    plt.imshow(cv2.cvtColor(image_2, cv2.COLOR_BGR2RGB))
    plt.title(title_2)
    plt.show()
```

```python
image = cv2.imread("IMG_0259.jpg")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
```

```python
rows, cols,_ = image.shape
noise = np.random.normal(0,30,(rows, cols, 3)).astype(np.uint8)
noisy_image = image + noise
plot_image(image, noisy_image, title_2="Image Plus Noise")
```

### Filtering Noise

```python
kernel = np.ones((10,10))/100
image_filtered = cv2.filter2D(src=noisy_image, ddepth=-1, kernel=kernel)
plot_image(noisy_image, image_filtered, title_1="Noisy Image", title_2="Filtered Image")
```

### Gaussian Blur

```python
image_filtered = cv2.GaussianBlur(noisy_image, (5,5),sigmaX=8, sigmaY=8)
plot_image(noisy_image, image_filtered, title_1="Noisy Image", title_2="Filtered Image")
```

```python
image_filtered = cv2.GaussianBlur(noisy_image, (3,5),sigmaX=8, sigmaY=8)
plot_image(noisy_image, image_filtered, title_1="Noisy Image", title_2="Filtered Image")
```

### Edge Detection

Edges are where pixel intensity changes. Edge is detected using the gradient of an unfiltered image. There are several methods to approximate the gradient, like the Sobel Edge Detector. 

```python
ddepth=cv2.CV_16S
grad_X = cv2.Sobel(src=image, ddepth=ddepth, dx=1, dy=1, ksize=5)
```
